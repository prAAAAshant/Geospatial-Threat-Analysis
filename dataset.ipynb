```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
```

```python
df = pd.read_csv('wildfire_dataset.csv')
```

```python
print("Raw Dataset:")
print(df.head())
```

```python
columns_to_drop = ['satellite', 'instrument', 'confidence', 'version', 'daynight']
df = df.drop(columns=columns_to_drop)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df['acq_date'] = pd.to_datetime(df['acq_date'])
```

```python
df['year'] = df['acq_date'].dt.year
df['month'] = df['acq_date'].dt.month
df['day'] = df['acq_date'].dt.day
```

```python
print("\nCleaned Dataset:")
print(df.head())
```

```python
df.to_csv('wildfire_data_cleaned.csv', index=False)
print("\nCleaned dataset saved as 'wildfire_data_cleaned.csv'.")
```
